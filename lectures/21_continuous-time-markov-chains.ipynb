```python
# | echo: false

import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(151)
plt.rcParams["figure.figsize"] = (6, 4)
```

# Continuous-time Markov chains

::: {.callout-note icon=false}
## Definition: CTMC
The process stays in state $i$ for an **Exponential**$(q_i)$ holding time (memoryless), then jumps to $j\neq i$ w.p. $q_{ij}/q_i$, where $q_i=\sum_{j\neq i}q_{ij}$. The rate matrix $Q$ plays the role $P$ played in discrete time.
:::

. . .

::: {.callout-warning icon=false}
## Fact
Long-run behavior is again governed by a stationary distribution, satisfying $\pi Q = 0$.
:::

---

## Simulating a CTMC via the next-event method

```python
q01, q10 = 0.5, 1.0
T = 200.0
t, state = 0.0, 0
times, states = [0.0], [0]
while t < T:
    rate = q01 if state == 0 else q10
    t += rng.exponential(1 / rate)
    state = 1 - state
    times.append(t); states.append(state)

times, states = np.array(times), np.array(states)
plt.step(times[:60], states[:60], where="post")
plt.yticks([0, 1]); plt.xlabel("t"); plt.title("CTMC sample path")
plt.show()
```

```python
durations = np.diff(times)
time_in_state0 = durations[states[:-1] == 0].sum()
print("empirical fraction of time in state 0:", round(time_in_state0 / times[-1], 3))
print("theoretical pi_0 = q10/(q01+q10):", round(q10 / (q01 + q10), 3))
```

We now shift from *modeling* randomness to *estimating* things from it.
