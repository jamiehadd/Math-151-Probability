# From PMFs to PDFs

::: {.callout-note icon=false}
## Definition: PDF and CDF
A continuous r.v. $X$ has PDF $f_X \geq 0$ with $\int_{-\infty}^\infty f_X = 1$, and $$P(a\le X\le b) = \int_a^b f_X(x)\,dx.$$ The CDF is $F_X(x)=P(X\le x)=\int_{-\infty}^x f_X(t)\,dt$, with $f_X = F_X'$ where differentiable.
:::

. . .

::: {.callout-warning icon=false}
## Fact
$E[X] = \int x f_X(x)\,dx$, &nbsp; $\operatorname{Var}(X) = \int (x-E[X])^2 f_X(x)\,dx$.
:::

## Uniform and Exponential: theory vs. simulation

```python
from scipy import stats

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
xs = np.linspace(-0.5, 5, 400)
for ax, dist, name in [(axes[0], stats.uniform(0, 2), "Uniform(0,2)"),
                        (axes[1], stats.expon(scale=1.0), "Exponential(rate=1)")]:
    ax.plot(xs, dist.pdf(xs), label="theoretical PDF")
    samples = dist.rvs(size=20000, random_state=rng)
    ax.hist(samples, bins=50, density=True, alpha=0.5, label="simulated histogram")
    ax.set_title(name); ax.legend()
plt.tight_layout(); plt.show()
```
