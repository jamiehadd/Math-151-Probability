# Moment generating functions

::: {.callout-note icon=false}
## Definition: MGF
$$M_X(t) = E\big[e^{tX}\big].$$ Differentiating and setting $t=0$ recovers moments: $M_X'(0)=E[X]$, $M_X''(0)=E[X^2]$, etc.
:::

. . .

::: {.callout-warning icon=false}
## Fact
If $X,Y$ independent, $M_{X+Y}(t) = M_X(t)M_Y(t)$. Two r.v.'s with the same MGF near $0$ have the same distribution.
:::

---

## MGF of a Normal, empirically

```python
mu, sigma = 0, 1
X = rng.normal(mu, sigma, size=200000)
ts = np.linspace(-1, 1, 50)
empirical_mgf = [np.mean(np.exp(t * X)) for t in ts]
theoretical_mgf = np.exp(mu * ts + 0.5 * sigma**2 * ts**2)

plt.plot(ts, empirical_mgf, "o", markersize=3, label="empirical")
plt.plot(ts, theoretical_mgf, label=r"$e^{\mu t + \sigma^2 t^2/2}$")
plt.legend(); plt.xlabel("t"); plt.title("MGF of a Normal(0,1)")
plt.show()
```

# Case Study: A Sub-Gaussian Bound for Rademacher Variables

::: {.callout-warning icon=false}
## Fact
A Rademacher $z\in\{-1,+1\}$ has $M_z(t) = \cosh(t) \le e^{t^2/2}$ for all $t$ -- exactly the algebraic fact that proves concentration for the trace estimator in three lectures.
:::

```python
ts = np.linspace(-3, 3, 200)
plt.plot(ts, np.cosh(ts), label=r"$M_z(t)=\cosh(t)$")
plt.plot(ts, np.exp(ts**2 / 2), "--", label=r"$e^{t^2/2}$")
plt.legend(); plt.xlabel("t"); plt.title(r"$\cosh(t) \leq e^{t^2/2}$")
plt.show()
```
