# Covariance and correlation

::: {.callout-note icon=false}
## Definition: Covariance / correlation
$$\operatorname{Cov}(X,Y) = E[XY]-E[X]E[Y], \qquad \rho_{X,Y} = \frac{\operatorname{Cov}(X,Y)}{\sigma_X\sigma_Y}\in[-1,1].$$ Independence $\Rightarrow \operatorname{Cov}=0$, but not conversely.
:::

. . .

::: {.callout-warning icon=false}
## Fact
For **independent** $X_1,\ldots,X_m$: $\operatorname{Var}(\sum_i X_i) = \sum_i \operatorname{Var}(X_i)$ (cross terms vanish).
:::

---

::: {.callout-warning icon=false}
## Theorem: Laws of Total Expectation and Variance
$$E[X] = E\big[E[X\mid Y]\big], \qquad \operatorname{Var}(X) = E\big[\operatorname{Var}(X\mid Y)\big] + \operatorname{Var}\big(E[X\mid Y]\big).$$
:::

## Averaging reduces variance

```python
sigma2 = 4.0
ms = [1, 2, 4, 8, 16, 32, 64, 128]
empirical_var = [np.var([rng.normal(0, np.sqrt(sigma2), size=m).mean() for _ in range(5000)]) for m in ms]

plt.loglog(ms, empirical_var, "o-", label="empirical Var(mean of m samples)")
plt.loglog(ms, sigma2 / np.array(ms), "k--", label=r"$\sigma^2/m$")
plt.xlabel("m"); plt.ylabel("variance"); plt.legend()
plt.title("Independence -> variance of the average shrinks like 1/m")
plt.show()
```

# Case Study: Variance Reduction for Trace Estimation

Averaging $m$ *independent* copies of Hutchinson's estimator, $\bar T = \frac1m\sum_k T^{(k)}$, gives $\operatorname{Var}(\bar T) = \operatorname{Var}(T)/m$ -- today's result, applied to our running example.

```python
n = 50
A = rng.normal(size=(n, n))

def hutchinson_average(m):
    total = 0.0
    for _ in range(m):
        z = rng.choice([-1, 1], size=n)
        total += z @ A @ z
    return total / m

ms = [1, 2, 4, 8, 16, 32, 64]
var_by_m = [np.var([hutchinson_average(m) for _ in range(1000)]) for m in ms]

plt.loglog(ms, var_by_m, "o-", label=r"empirical Var($\bar T$)")
plt.loglog(ms, var_by_m[0] / np.array(ms), "k--", label="Var(single)/m")
plt.xlabel("m (# samples averaged)"); plt.ylabel("variance"); plt.legend()
plt.title("Trace estimator variance shrinks like 1/m")
plt.show()
```
