# Convergence in probability

::: {.callout-note icon=false}
## Definition: Convergence in probability
$Y_n \xrightarrow{p} c$ if for every $\varepsilon>0$, $$P(|Y_n-c|\ge\varepsilon)\to 0.$$ The WLLN is exactly the statement $\bar X_n \xrightarrow{p}\mu$.
:::

---

::: {.callout-warning icon=false}
## Theorem: Central Limit Theorem
For iid $X_1,\ldots,X_n$ with mean $\mu$, variance $\sigma^2$: $$\frac{\bar X_n - \mu}{\sigma/\sqrt n} \xrightarrow{d} N(0,1).$$ For large $n$, $\bar X_n \approx N(\mu,\sigma^2/n)$ -- regardless of the distribution of the $X_i$.
:::

## Watching the CLT happen

```python
from scipy.stats import norm
fig, axes = plt.subplots(1, 3, figsize=(12, 3.5))
for ax, n in zip(axes, [1, 5, 30]):
    means = rng.exponential(1.0, size=(20000, n)).mean(axis=1)
    ax.hist(means, bins=50, density=True, alpha=0.6)
    xs = np.linspace(means.min(), means.max(), 200)
    ax.plot(xs, norm.pdf(xs, 1, 1/np.sqrt(n)), "k--")
    ax.set_title(f"n = {n}")
plt.suptitle("CLT: average of n Exponential(1) variables")
plt.tight_layout(); plt.show()
```

# Case Study: Kaczmarz Iterates Converge in Probability

Randomized Kaczmarz satisfies $E\big[\|x_k-x^\star\|^2\big] \le \rho^k\|x_0-x^\star\|^2$ for some $\rho<1$. Combined with Markov's inequality (L14), this implies $x_k \xrightarrow{p} x^\star$.

```python
m, n = 40, 10
A = rng.normal(size=(m, n))
x_true = rng.normal(size=n)
b = A @ x_true
p = np.sum(A**2, axis=1); p /= p.sum()

def kaczmarz_error_at(k):
    x = np.zeros(n)
    for _ in range(k):
        i = rng.choice(m, p=p)
        a_i, b_i = A[i], b[i]
        x = x + (b_i - a_i @ x) / (a_i @ a_i) * a_i
    return np.linalg.norm(x - x_true)

k = 60
errors_at_k = [kaczmarz_error_at(k) for _ in range(2000)]
plt.hist(errors_at_k, bins=40)
plt.xlabel(f"||x_{{{k}}} - x*||")
plt.title("Kaczmarz error distribution: mass concentrating near 0")
plt.show()
print("initial ||x_0 - x*|| =", round(np.linalg.norm(x_true), 3))
print(f"mean ||x_{{{k}}} - x*|| =", round(np.mean(errors_at_k), 3), " (most of the mass has moved near 0)")
```
