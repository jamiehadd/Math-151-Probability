# Stationary distributions

::: {.callout-note icon=false}
## Definition: Stationary distribution
A distribution $\pi$ is **stationary** for $P$ if $\pi P = \pi$. For irreducible, aperiodic chains, $\pi$ exists, is unique, and $P(X_n=i)\to\pi_i$ regardless of the start.
:::

. . .

::: {.callout-warning icon=false}
## Fact
This connects back to the LLN: the long-run fraction of time in state $i$ converges (in probability) to $\pi_i$.
:::

---

## Computing and verifying a stationary distribution

```python
P = np.array([[0.8, 0.2], [0.4, 0.6]])
A_eq = np.vstack([P.T - np.eye(2), np.ones(2)])
b_eq = np.array([0, 0, 1])
pi, *_ = np.linalg.lstsq(A_eq, b_eq, rcond=None)
print("stationary distribution pi =", np.round(pi, 4))

x, history = 0, []
for _ in range(200000):
    history.append(x)
    x = rng.choice(2, p=P[x])
history = np.array(history)
print("empirical long-run fractions:", np.round([np.mean(history==0), np.mean(history==1)], 4))
```
