# The MMSE estimator

::: {.callout-note icon=false}
## Definition: Bayesian MMSE
Among *all* estimators, the one minimizing $E[(\theta-\hat\theta)^2]$ is the posterior mean: $$\hat\theta_{\text{MMSE}} = E[\theta\mid\text{data}].$$
:::

. . .

::: {.callout-warning icon=false}
## Theorem: Optimality of the conditional expectation
$E[\theta\mid\text{data}]$ minimizes $E[(\theta-c)^2\mid\text{data}]$ over all constants $c$ (given the data) -- the same fact underlying the law of total variance (Lecture 10).
:::

## MMSE vs. MAP vs. MLE, side by side

```python
from scipy.stats import beta as beta_dist

theta_true, n_flips = 0.65, 15
prior_a, prior_b = 2, 2
flips = rng.random(n_flips) < theta_true
k = flips.sum()
post_a, post_b = prior_a + k, prior_b + (n_flips - k)

mle = k / n_flips
map_est = (post_a - 1) / (post_a + post_b - 2)
mmse = post_a / (post_a + post_b)

thetas = np.linspace(0.001, 0.999, 400)
plt.plot(thetas, beta_dist.pdf(thetas, post_a, post_b), label="posterior")
plt.axvline(mle, color="C1", linestyle="--", label=f"MLE = {mle:.3f}")
plt.axvline(map_est, color="C2", linestyle=":", label=f"MAP = {map_est:.3f}")
plt.axvline(mmse, color="C3", linestyle="-.", label=f"MMSE = {mmse:.3f}")
plt.axvline(theta_true, color="black", label="truth")
plt.legend(); plt.title("Three point estimates from the same posterior")
plt.show()
```
