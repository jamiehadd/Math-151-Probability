# Finding the distribution of Y = g(X)

::: {.callout-warning icon=false}
## Fact
**CDF method:** compute $F_Y(y) = P(g(X)\le y)$ directly.
:::

. . .

::: {.callout-warning icon=false}
## Theorem: Density transformation formula
For monotonic, differentiable $g$: $$f_Y(y) = f_X\big(g^{-1}(y)\big)\left|\frac{d}{dy}g^{-1}(y)\right|.$$
:::

## Worked example: Y = X^2 for X ~ N(0,1)

```python
from scipy.stats import chi2

X = rng.normal(size=200000)
Y = X**2
ys = np.linspace(0.001, 8, 400)

plt.hist(Y, bins=80, density=True, range=(0, 8), alpha=0.5, label="simulated Y=X^2")
plt.plot(ys, chi2.pdf(ys, df=1), label=r"derived PDF ($\chi^2_1$)")
plt.legend(); plt.xlabel("y"); plt.title("Deriving the distribution of Y = X^2")
plt.show()
```

# Case Study: Randomized Matrix Multiplication

Write $AB = \sum_{i=1}^m a_ib_i^\top$ (a sum of rank-one terms, columns $a_i$ of $A$ / rows $b_i^\top$ of $B$). Sample a single index $I$ and rescale:

::: {.callout-note icon=false}
## Definition: Sampling estimator for AB
$$Z = \frac{1}{p_I}a_Ib_I^\top, \qquad p_i \propto \|a_i\|\,\|b_i\|.$$ Unlike a named family, $p$ is **derived** directly from the data. We'll find the *best* choice of $p$ in Week 9 (Lecture 16).
:::

```python
n, m, p_dim = 6, 40, 5
A = rng.normal(size=(n, m))
B = rng.normal(size=(m, p_dim))
true_AB = A @ B

norms = np.linalg.norm(A, axis=0) * np.linalg.norm(B, axis=1)
p = norms / norms.sum()

def one_sample_estimator():
    i = rng.choice(m, p=p)
    return np.outer(A[:, i], B[i, :]) / p[i]

avg_estimate = np.mean([one_sample_estimator() for _ in range(4000)], axis=0)
print("max entrywise error after averaging 4000 single-sample estimates:")
print(np.max(np.abs(avg_estimate - true_AB)))
```
