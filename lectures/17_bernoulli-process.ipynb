# The Bernoulli process

::: {.callout-note icon=false}
## Definition: Bernoulli process
A sequence $X_1,X_2,\ldots$ of iid Bernoulli($p$) trials. Let $T$ = time of first success.
:::

. . .

::: {.callout-warning icon=false}
## Theorem: Distribution of T
$$P(T=k) = (1-p)^{k-1}p, \quad k=1,2,\ldots \quad (\text{Geometric}(p)), \qquad E[T]=\frac1p.$$
:::

---

::: {.callout-warning icon=false}
## Fact
**Memorylessness:** $P(T>s+t\mid T>s) = P(T>t)$ for integers $s,t\ge0$ -- waiting doesn't help or hurt.
:::

## Simulating a Bernoulli process

```python
from scipy.stats import geom
p = 0.15
trials = rng.random(4000) < p
success_times = np.where(trials)[0]
interarrivals = np.diff(success_times)

values, counts = np.unique(interarrivals, return_counts=True)
plt.bar(values, counts / counts.sum(), alpha=0.6, label="empirical interarrivals")
plt.plot(values, geom.pmf(values, p), "ko-", label="Geometric(p) PMF")
plt.legend(); plt.xlim(0, 40); plt.title("Bernoulli process interarrival times")
plt.show()
```

```python
s, t = 3, 5
print(f"P(T > s+t | T > s) ~ {np.mean(interarrivals[interarrivals > s] - s > t):.3f}")
print(f"P(T > t)           ~ {np.mean(interarrivals > t):.3f}")
```
