# Bayesian estimation

::: {.callout-note icon=false}
## Definition: Prior, posterior, MAP
The parameter $\theta$ has a **prior** $f(\theta)$, updated via Bayes' rule (L8) into a **posterior** $f(\theta\mid\text{data}) \propto L(\text{data}\mid\theta)f(\theta)$. The **MAP estimator** is $$\hat\theta_{\text{MAP}} = \arg\max_\theta\big[\log L(\text{data}\mid\theta) + \log f(\theta)\big].$$
:::

. . .

::: {.callout-warning icon=false}
## Fact
Compare to the MLE (L24), which drops the prior term. MAP and MLE agree as $n$ grows and the prior's influence washes out.
:::

## MAP vs. MLE for a coin's bias

```python
from scipy.stats import beta as beta_dist

theta_true = 0.65
prior_a, prior_b = 2, 2

fig, axes = plt.subplots(1, 3, figsize=(13, 3.5))
for ax, n_flips in zip(axes, [5, 20, 200]):
    flips = rng.random(n_flips) < theta_true
    k = flips.sum()
    post_a, post_b = prior_a + k, prior_b + (n_flips - k)

    thetas = np.linspace(0.001, 0.999, 400)
    mle = k / n_flips
    map_est = (post_a - 1) / (post_a + post_b - 2)

    ax.plot(thetas, beta_dist.pdf(thetas, post_a, post_b), label="posterior")
    ax.axvline(mle, color="C1", linestyle="--", label=f"MLE={mle:.2f}")
    ax.axvline(map_est, color="C2", linestyle=":", label=f"MAP={map_est:.2f}")
    ax.axvline(theta_true, color="black", label="truth")
    ax.set_title(f"n = {n_flips}"); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
```
