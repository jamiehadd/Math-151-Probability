# Beyond convergence in probability

::: {.callout-warning icon=false}
## Theorem: Strong Law of Large Numbers
If $X_1,X_2,\ldots$ iid mean $\mu$: $\bar X_n \to \mu$ **almost surely**, i.e. $P(\lim_n \bar X_n = \mu) = 1$ -- stronger than convergence in probability.
:::

---

::: {.callout-warning icon=false}
## Theorem: Chernoff Bound
For $t>0$, Markov applied to $e^{tX}$ gives $P(X\ge a) \le e^{-ta}M_X(t)$; optimizing over $t$: $$P(X\ge a) \le \min_{t>0} e^{-ta}M_X(t),$$ typically decaying *exponentially* -- far tighter than Chebyshev.
:::

## Chebyshev vs. Chernoff, head to head

```python
n = 100
S = rng.choice([-1, 1], size=(200000, n)).sum(axis=1)
a_values = np.arange(5, 45, 5)
true_probs = [np.mean(S >= a) for a in a_values]
chebyshev = [min(n / a**2, 1) for a in a_values]
chernoff = [np.exp(-a**2 / (2 * n)) for a in a_values]

plt.semilogy(a_values, true_probs, "o-", label="true P(S >= a)")
plt.semilogy(a_values, chebyshev, "s--", label="Chebyshev bound")
plt.semilogy(a_values, chernoff, "^--", label="Chernoff bound")
plt.xlabel("a"); plt.ylabel("probability (log)"); plt.legend()
plt.title("Chernoff is much tighter in the tail")
plt.show()
```

# Case Study: Optimal Sampling for Randomized Matrix Multiplication

Back in Week 5 we sampled $p_i \propto \|a_i\|\|b_i\|$ without justifying it. One can show (Cauchy-Schwarz, minimizing variance over the simplex) this is exactly the **variance-minimizing** choice -- an optimization-meets-probability moment -- and it also gives sharper concentration than uniform sampling.

```python
n_, m, p_dim = 6, 60, 5
A = rng.normal(size=(n_, m))
B = rng.normal(size=(m, p_dim))
true_AB = A @ B

def sample_error(p, n_trials=800):
    errs = []
    for _ in range(n_trials):
        i = rng.choice(m, p=p)
        est = np.outer(A[:, i], B[i, :]) / p[i]
        errs.append(np.linalg.norm(est - true_AB))
    return np.array(errs)

p_uniform = np.full(m, 1 / m)
norms = np.linalg.norm(A, axis=0) * np.linalg.norm(B, axis=1)
p_optimal = norms / norms.sum()

errs_u, errs_o = sample_error(p_uniform), sample_error(p_optimal)
plt.hist(errs_u, bins=40, alpha=0.5, label=f"uniform (std={errs_u.std():.2f})")
plt.hist(errs_o, bins=40, alpha=0.5, label=f"optimal (std={errs_o.std():.2f})")
plt.legend(); plt.xlabel("single-sample error")
plt.title("Optimal sampling has much lower spread")
plt.show()
```
