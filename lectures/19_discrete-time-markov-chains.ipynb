# Markov chains

::: {.callout-note icon=false}
## Definition: Markov property
A process $X_0,X_1,\ldots$ has the **Markov property** if $$P(X_{n+1}=j\mid X_n=i,X_{n-1},\ldots,X_0) = P(X_{n+1}=j\mid X_n=i) = p_{ij}$$ -- the future depends on the past only through the present.
:::

. . .

::: {.callout-warning icon=false}
## Fact
The $p_{ij}$ form the **transition matrix** $P$; $n$-step transition probabilities are given by $P^n$.
:::

---

## A simple weather chain

```python
P = np.array([[0.8, 0.2], [0.4, 0.6]])
x = 0
history = [x]
for _ in range(2000):
    x = rng.choice(2, p=P[x])
    history.append(x)

print("empirical fraction sunny:", np.mean(np.array(history) == 0))
print("P^10 =\n", np.round(np.linalg.matrix_power(P, 10), 3))
```

# Case Study: Randomized Kaczmarz as a Markov Chain

$x_{k+1}$ is built from $x_k$ and a *fresh, independent* random row draw -- it does not depend on $x_{k-1},x_{k-2},\ldots$ once $x_k$ is known. That's the Markov property, on the continuous state space $\mathbb{R}^n$.

```python
m, n = 20, 5
A = rng.normal(size=(m, n))
x_true = rng.normal(size=n)
b = A @ x_true
p = np.sum(A**2, axis=1); p /= p.sum()

def kaczmarz_step(x):
    i = rng.choice(m, p=p)
    a_i, b_i = A[i], b[i]
    return x + (b_i - a_i @ x) / (a_i @ a_i) * a_i

x_k = rng.normal(size=n)  # a fixed "current" iterate, from two different histories
next_A = np.array([kaczmarz_step(x_k) for _ in range(4000)])
next_B = np.array([kaczmarz_step(x_k) for _ in range(4000)])
print("mean of x_(k+1) | history A:", np.round(next_A.mean(axis=0), 3))
print("mean of x_(k+1) | history B:", np.round(next_B.mean(axis=0), 3))
print("-> matches regardless of history: the Markov property holds.")
```
