# Bayes' rule for random variables

::: {.callout-warning icon=false}
## Theorem: Bayes' Rule
For jointly distributed $X,Y$: $$f_{X\mid Y}(x\mid y) = \frac{f_{Y\mid X}(y\mid x)\,f_X(x)}{f_Y(y)}.$$ This holds even when one variable is discrete and the other continuous.
:::

. . .

**Example:** A coin has unknown bias $\Theta \sim \text{Uniform}(0,1)$ (our *prior*). We observe $k$ heads in $n$ flips. The *posterior* is $$f_{\Theta\mid\text{data}}(\theta) \propto \theta^k(1-\theta)^{n-k},$$ a $\text{Beta}(k+1,n-k+1)$ density -- we'll meet this machinery again in Week 15.

---

## Updating a belief about a coin's bias

```python
from scipy.stats import beta

theta_true = 0.7
n_flips = 30
flips = rng.random(n_flips) < theta_true
k = flips.sum()

thetas = np.linspace(0, 1, 400)
prior = np.ones_like(thetas)
posterior = beta.pdf(thetas, k + 1, n_flips - k + 1)

plt.plot(thetas, prior, label="prior")
plt.plot(thetas, posterior, label=f"posterior ({k}/{n_flips} heads)")
plt.axvline(theta_true, color="black", linestyle="--", label="true bias")
plt.xlabel(r"$\theta$"); plt.legend(); plt.title("Bayes' rule updates our belief")
plt.show()
```
