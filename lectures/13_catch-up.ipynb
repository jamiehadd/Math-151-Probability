# Where we are with our running examples

- **Randomized Kaczmarz** (L3): a distribution $p_i \propto \|a_i\|^2$ built from data; error decreases geometrically *in expectation*.
- **Randomized trace estimation** (L4, L10, L11): an unbiased estimator via linearity; variance shrinks like $1/m$ when averaging; Rademacher MGF is sub-Gaussian.
- **Randomized matrix multiplication** (L9): another unbiased, derived-distribution estimator, same shape as trace estimation.

. . .

::: {.callout-warning icon=false}
## Fact
All three are *unbiased randomized estimators* -- the tools we've built (linearity, variance, MGFs) apply to all of them. That's the whole point of running examples.
:::

## Side-by-side: two unbiased estimators

```python
n = 15
A = rng.normal(size=(n, n))
true_trace = np.trace(A)
zs = rng.choice([-1, 1], size=(3000, n))
trace_ests = np.einsum("ki,ij,kj->k", zs, A, zs)

B = rng.normal(size=(n, n)); C = rng.normal(size=(n, n))
true_prod = B @ C
norms = np.linalg.norm(B, axis=0) * np.linalg.norm(C, axis=1)
p = norms / norms.sum()
idx = rng.choice(n, size=3000, p=p)
matmul_ests = [np.outer(B[:, i], C[i, :])[0, 0] / p[i] for i in idx]

print(f"Trace estimator:  true={true_trace:.3f}, sample mean={trace_ests.mean():.3f}")
print(f"Matmul estimator: true (0,0)={true_prod[0,0]:.3f}, sample mean={np.mean(matmul_ests):.3f}")
```
