# Concentration inequalities

::: {.callout-warning icon=false}
## Theorem: Markov's Inequality
For $X\ge 0$, $a>0$: $$P(X\ge a) \le \frac{E[X]}{a}.$$
:::

. . .

::: {.callout-warning icon=false}
## Theorem: Chebyshev's Inequality
Applying Markov to $(X-\mu)^2$: for $k>0$, $$P(|X-\mu|\ge k) \le \frac{\sigma^2}{k^2}.$$
:::

---

::: {.callout-warning icon=false}
## Theorem: Weak Law of Large Numbers
If $X_1,X_2,\ldots$ iid with mean $\mu$, variance $\sigma^2$, and $\bar X_n = \frac1n\sum X_i$, then for every $\epsilon>0$, $$P(|\bar X_n-\mu|\ge\epsilon)\to 0.$$
:::

<details><summary>Proof:</summary>
$\operatorname{Var}(\bar X_n)=\sigma^2/n\to 0$, so Chebyshev gives $P(|\bar X_n-\mu|\ge\epsilon)\le \sigma^2/(n\epsilon^2)\to 0$.
</details>

## Chebyshev's bound vs. reality

```python
mu, sigma2 = 5, 9
n = 30
means = np.array([rng.normal(mu, np.sqrt(sigma2), size=n).mean() for _ in range(20000)])

for eps in [0.5, 1.0, 1.5, 2.0]:
    empirical = np.mean(np.abs(means - mu) >= eps)
    bound = min(sigma2 / n / eps**2, 1)
    print(f"eps={eps:.1f}:  empirical P = {empirical:.4f}   Chebyshev bound = {bound:.4f}")
```

# Case Study: A Concentration Guarantee for Trace Estimation

$\operatorname{Var}(\bar T) = \operatorname{Var}(T)/m$ combined with Chebyshev gives a sample-size guarantee: to ensure $P(|\bar T-\operatorname{tr}(A)|\ge\varepsilon)\le\delta$, take $$m \ge \frac{\operatorname{Var}(T)}{\varepsilon^2\delta}.$$

```python
n = 60
A = rng.normal(size=(n, n))
true_trace = np.trace(A)

z_single = rng.choice([-1, 1], size=(4000, n))
var_T = np.einsum("ki,ij,kj->k", z_single, A, z_single).var()

eps, delta = 5.0, 0.1
m_needed = int(np.ceil(var_T / (eps**2 * delta)))
print(f"Var(T)={var_T:.2f} -> Chebyshev says m >= {m_needed}")

def trace_average(m):
    zs = rng.choice([-1, 1], size=(m, n))
    return np.einsum("ki,ij,kj->k", zs, A, zs).mean()

trials = np.array([trace_average(m_needed) for _ in range(300)])
print(f"empirical P(error >= {eps}): {np.mean(np.abs(trials - true_trace) >= eps):.3f}  (promised <= {delta})")
```
