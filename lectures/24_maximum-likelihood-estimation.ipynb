# Maximum likelihood

::: {.callout-note icon=false}
## Definition: Likelihood / MLE
For data $x_1,\ldots,x_n$ with parameter $\theta$: $L(\theta)=\prod_i f(x_i;\theta)$, and $$\hat\theta_{\text{MLE}} = \arg\max_\theta L(\theta) = \arg\max_\theta \log L(\theta).$$
:::

. . .

::: {.callout-warning icon=false}
## Fact
**Poisson example:** $\log L(\lambda)=\sum_i(x_i\log\lambda-\lambda-\log x_i!)$; setting the derivative to zero gives $\hat\lambda_{\text{MLE}} = \bar x$.
:::

---

## Recovering lambda by maximizing the log-likelihood

```python
from scipy.optimize import minimize_scalar
from scipy.stats import poisson

lam_true, n = 4.5, 40
data = rng.poisson(lam_true, size=n)

def neg_log_likelihood(lam):
    return np.inf if lam <= 0 else -np.sum(poisson.logpmf(data, lam))

result = minimize_scalar(neg_log_likelihood, bounds=(0.01, 20), method="bounded")
print("numerical MLE:", round(result.x, 3))
print("closed-form MLE (sample mean):", round(data.mean(), 3))
```

```python
ns = [5, 20, 100, 500, 2000]
means_by_n = [np.mean([rng.poisson(lam_true, size=n_).mean() for _ in range(500)]) for n_ in ns]
stds_by_n = [np.std([rng.poisson(lam_true, size=n_).mean() for _ in range(500)]) for n_ in ns]

plt.errorbar(ns, means_by_n, yerr=stds_by_n, fmt="o-")
plt.axhline(lam_true, color="black", linestyle="--", label="true lambda")
plt.xscale("log"); plt.xlabel("n"); plt.legend()
plt.title("MLE concentrates near the truth as n grows")
plt.show()
```
