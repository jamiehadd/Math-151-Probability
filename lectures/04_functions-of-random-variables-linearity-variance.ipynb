# LOTUS and linearity

::: {.callout-warning icon=false}
## Theorem: Law of the Unconscious Statistician (LOTUS)
$$E[g(X)] = \sum_x g(x)\,p_X(x)$$ -- no need to first find the distribution of $g(X)$.
:::

. . .

::: {.callout-warning icon=false}
## Theorem: Linearity of Expectation
For *any* random variables $X_1,\ldots,X_n$ (independent or not!) and constants $a_i$, $$E\Big[\sum_i a_i X_i\Big] = \sum_i a_i E[X_i].$$
:::

---

::: {.callout-note icon=false}
## Definition: Variance
$$\operatorname{Var}(X) = E\big[(X-E[X])^2\big] = E[X^2]-(E[X])^2, \qquad \operatorname{Var}(aX+b)=a^2\operatorname{Var}(X).$$
:::

## Linearity in action: fixed points of a random permutation

Shuffle $n$ cards; let $X$ = number of cards in their original position. $X$ is a sum of *dependent* indicators, but linearity doesn't care!

```python
n = 52
n_trials = 20000
fixed_points = [np.sum(rng.permutation(n) == np.arange(n)) for _ in range(n_trials)]

print("empirical E[X]   =", np.mean(fixed_points), " (linearity: n * 1/n = 1)")
print("empirical Var(X) =", np.var(fixed_points), " (theory -> 1 as n grows)")
```

# Case Study: Randomized Trace Estimation

For $A \in \mathbb{R}^{n\times n}$, let $z$ be a **Rademacher** vector (entries iid $\pm1$). Hutchinson's estimator is $T = z^\top A z$.

::: {.callout-warning icon=false}
## Theorem: Unbiasedness of the trace estimator
$E[T] = \operatorname{tr}(A)$.
:::

<details><summary>Proof:</summary>
Using $E[z_iz_j]=0$ for $i\neq j$ and $E[z_i^2]=1$, linearity gives $$E[T] = E\Big[\sum_{i,j}z_iA_{ij}z_j\Big] = \sum_{i,j}A_{ij}E[z_iz_j] = \sum_i A_{ii} = \operatorname{tr}(A).$$
</details>

```python
n = 50
A = rng.normal(size=(n, n))
true_trace = np.trace(A)

def hutchinson_estimate():
    z = rng.choice([-1, 1], size=n)
    return z @ A @ z

estimates = np.array([hutchinson_estimate() for _ in range(5000)])
print("true trace(A)         =", round(true_trace, 3))
print("mean of estimator     =", round(estimates.mean(), 3))
print("empirical Var(single) =", round(estimates.var(), 3))

plt.hist(estimates, bins=40)
plt.axvline(true_trace, color="black", linestyle="--", label="true trace")
plt.legend(); plt.xlabel("single estimate  z^T A z"); plt.title("Hutchinson trace estimator")
plt.show()
```
