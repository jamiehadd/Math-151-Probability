# The Poisson process

::: {.callout-note icon=false}
## Definition: Poisson process
The continuous-time limit of a Bernoulli process, rate $\lambda$. $N(t)$ = number of arrivals by time $t$: $N(t)\sim\text{Poisson}(\lambda t)$, and interarrival times are iid $\text{Exponential}(\lambda)$ (memoryless).
:::

. . .

::: {.callout-warning icon=false}
## Fact
**Independent/stationary increments:** counts over disjoint intervals are independent, and their distribution depends only on interval length. **Merging:** independent Poisson($\lambda_1$), Poisson($\lambda_2$) processes union to Poisson($\lambda_1+\lambda_2$). **Splitting:** routing each arrival independently with prob. $q$ yields independent Poisson($\lambda q$), Poisson($\lambda(1-q)$).
:::

---

## Simulating a Poisson process

```python
lam, T = 3.0, 20.0
interarrivals = rng.exponential(1/lam, size=2000)
arrival_times = np.cumsum(interarrivals)
arrival_times = arrival_times[arrival_times < T]

ts = np.linspace(0, T, 500)
Nt = np.searchsorted(arrival_times, ts, side="right")
plt.step(ts, Nt, where="post")
plt.xlabel("t"); plt.ylabel("N(t)"); plt.title(f"Poisson process sample path (rate={lam})")
plt.show()
```

```python
from scipy.stats import poisson
counts = [np.sum(rng.exponential(1/lam, size=200).cumsum() < T) for _ in range(20000)]
values, freq = np.unique(counts, return_counts=True)
plt.bar(values, freq / freq.sum(), alpha=0.6, label="empirical N(T)")
plt.plot(values, poisson.pmf(values, lam * T), "ko-", label=f"Poisson({lam*T}) PMF")
plt.legend(); plt.title("N(T) is Poisson(lambda*T)"); plt.show()
```
