# Linear regression as estimation

::: {.callout-note icon=false}
## Definition: Simple linear regression
Model $Y_i = \beta_0+\beta_1x_i+\varepsilon_i$, $\varepsilon_i$ iid mean-zero. The least-squares estimator minimizes $\sum_i(y_i-\beta_0-\beta_1x_i)^2$: $$\hat\beta_1 = \frac{\sum_i(x_i-\bar x)(y_i-\bar y)}{\sum_i(x_i-\bar x)^2}, \qquad \hat\beta_0=\bar y-\hat\beta_1\bar x.$$
:::

. . .

::: {.callout-warning icon=false}
## Fact
If additionally $\varepsilon_i\sim N(0,\sigma^2)$, least squares is *exactly* the MLE for $(\beta_0,\beta_1)$ -- connecting directly back to Lecture 24.
:::

---

## Least squares == MLE under Gaussian noise

```python
n = 60
beta0_true, beta1_true, sigma = 2.0, 1.5, 1.0
x = rng.uniform(0, 10, size=n)
y = beta0_true + beta1_true * x + rng.normal(0, sigma, size=n)

X = np.column_stack([np.ones(n), x])
beta_ls, *_ = np.linalg.lstsq(X, y, rcond=None)
print("least-squares estimate (beta0, beta1):", np.round(beta_ls, 3))
```

```python
from scipy.optimize import minimize

def neg_log_lik(params):
    b0, b1, log_sigma = params
    sigma_ = np.exp(log_sigma)
    resid = y - (b0 + b1 * x)
    return np.sum(0.5 * (resid / sigma_) ** 2 + np.log(sigma_))

res = minimize(neg_log_lik, x0=[0, 0, 0])
print("MLE estimate (beta0, beta1):        ", np.round(res.x[:2], 3))

plt.scatter(x, y, s=15, alpha=0.6)
xs = np.linspace(0, 10, 100)
plt.plot(xs, beta_ls[0] + beta_ls[1] * xs, "r", label="fitted line")
plt.legend(); plt.title("Least squares fit"); plt.show()
```
