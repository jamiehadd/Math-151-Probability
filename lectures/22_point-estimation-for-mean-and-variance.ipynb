# Estimators, bias, variance, MSE

::: {.callout-note icon=false}
## Definition: Estimator, bias, MSE
An **estimator** $\hat\theta(X_1,\ldots,X_n)$ of $\theta$ is a random variable, with $$\operatorname{Bias}(\hat\theta)=E[\hat\theta]-\theta, \qquad \operatorname{MSE}(\hat\theta)=E[(\hat\theta-\theta)^2] = \operatorname{Bias}(\hat\theta)^2 + \operatorname{Var}(\hat\theta).$$
:::

. . .

::: {.callout-warning icon=false}
## Fact
$\bar X_n$ is an **unbiased** estimator of $\mu$. The sample variance $S_n^2 = \frac1{n-1}\sum_i(X_i-\bar X_n)^2$ (note $n-1$!) is unbiased for $\sigma^2$.
:::

## Bias correction: why n-1, not n?

```python
mu, sigma2, n = 2.0, 5.0, 8
biased, unbiased = [], []
for _ in range(30000):
    X = rng.normal(mu, np.sqrt(sigma2), size=n)
    xbar = X.mean()
    biased.append(np.sum((X - xbar)**2) / n)
    unbiased.append(np.sum((X - xbar)**2) / (n - 1))

print("true sigma^2            =", sigma2)
print("mean of /n estimator    =", round(np.mean(biased), 3), " (biased low)")
print("mean of /(n-1) estimator=", round(np.mean(unbiased), 3), " (unbiased)")
```

# Case Study: Randomized Matrix Multiplication as an Estimator

$Z = \frac1{p_I}a_Ib_I^\top$ is unbiased for $AB$, so $\operatorname{MSE}(Z)=\operatorname{Var}(Z)$. The optimal-sampling distribution from L16 is literally the **minimum-MSE** unbiased estimator in this family.

```python
n_, m, p_dim = 6, 50, 5
A = rng.normal(size=(n_, m))
B = rng.normal(size=(m, p_dim))
true_AB = A @ B

def mse(p, n_trials=400):
    sq = []
    for _ in range(n_trials):
        i = rng.choice(m, p=p)
        est = np.outer(A[:, i], B[i, :]) / p[i]
        sq.append(np.sum((est - true_AB) ** 2))
    return np.mean(sq)

p_uniform = np.full(m, 1 / m)
norms = np.linalg.norm(A, axis=0) * np.linalg.norm(B, axis=1)
p_optimal = norms / norms.sum()

print("MSE, uniform sampling:", round(mse(p_uniform), 3))
print("MSE, optimal sampling:", round(mse(p_optimal), 3))
```
