# Discrete random variables

::: {.callout-note icon=false}
## Definition: Random variable / PMF
A **random variable** $X$ is a function from $\Omega$ to $\mathbb{R}$. It is **discrete** if it takes finitely or countably many values, with **PMF** $p_X(x) = P(X=x)$, $\sum_x p_X(x) = 1$.
:::

. . .

::: {.callout-note icon=false}
## Definition: Expectation
$$E[X] = \sum_x x\,p_X(x),$$ the probability-weighted average, i.e. the long-run average value of $X$.
:::

---

::: {.callout-warning icon=false}
## Fact
**Bernoulli($p$):** $p_X(1)=p$, $E[X]=p$. &nbsp; **Binomial($n,p$):** $p_X(k) = \binom{n}{k}p^k(1-p)^{n-k}$, $E[X]=np$.
:::

## PMFs in practice

```python
from math import comb
n, p = 20, 0.3
samples = rng.binomial(n, p, size=100000)
values, counts = np.unique(samples, return_counts=True)
empirical_pmf = counts / counts.sum()
theoretical_pmf = [comb(n, k) * p**k * (1-p)**(n-k) for k in values]

plt.bar(values, empirical_pmf, alpha=0.6, label="empirical")
plt.plot(values, theoretical_pmf, "ko-", label="theoretical PMF")
plt.xlabel("k"); plt.ylabel("P(X = k)"); plt.legend()
plt.title(f"Binomial({n}, {p}): empirical vs. theoretical")
plt.show()
print("empirical E[X]  =", samples.mean(), " theoretical =", n*p)
```

# Case Study: The Randomized Kaczmarz Method

Given $Ax=b$ with rows $a_1,\ldots,a_m$, randomized Kaczmarz repeatedly picks a random row index $I$ and projects onto $\{x : a_I^\top x = b_I\}$.

::: {.callout-note icon=false}
## Definition: Row-sampling distribution
The row $I$ is sampled with $$p_i = \frac{\|a_i\|^2}{\|A\|_F^2}, \qquad i=1,\ldots,m,$$ where $\|A\|_F^2 = \sum_i \|a_i\|^2$ -- a genuine PMF over $\{1,\ldots,m\}$, built from the data in $A$.
:::

. . .

```python
m, n = 8, 4
A = rng.normal(size=(m, n))
x_true = rng.normal(size=n)
b = A @ x_true

row_norms_sq = np.sum(A**2, axis=1)
p = row_norms_sq / row_norms_sq.sum()
print("row-sampling PMF p_i:", np.round(p, 3), " sum =", p.sum())

draws = rng.choice(m, size=200000, p=p)
empirical = np.bincount(draws, minlength=m) / len(draws)
print("empirical frequencies:", np.round(empirical, 3))
```

```python
x = np.zeros(n)
errors = []
for k in range(200):
    i = rng.choice(m, p=p)
    a_i, b_i = A[i], b[i]
    x = x + (b_i - a_i @ x) / (a_i @ a_i) * a_i
    errors.append(np.linalg.norm(x - x_true))

plt.semilogy(errors)
plt.xlabel("iteration k"); plt.ylabel(r"$\|x_k - x^\star\|$ (log scale)")
plt.title("Randomized Kaczmarz: error decreasing over iterations")
plt.show()
```
