# Conditional probability

::: {.callout-note icon=false}
## Definition: Conditional probability
For events $A,B$ with $P(B)>0$, $$P(A \mid B) = \frac{P(A\cap B)}{P(B)}.$$
:::

. . .

::: {.callout-warning icon=false}
## Fact
**Multiplication rule:** $P(A \cap B) = P(B)P(A\mid B)$, extending to chains of events.
:::

---

::: {.callout-warning icon=false}
## Theorem: Law of Total Probability
If $B_1,\ldots,B_n$ partition $\Omega$, then for any event $A$, $$P(A) = \sum_{i=1}^n P(B_i)\,P(A\mid B_i).$$
:::

## Independence

::: {.callout-note icon=false}
## Definition: Independent events
$A$ and $B$ are **independent** if $P(A\cap B) = P(A)P(B)$, equivalently $P(A\mid B) = P(A)$.
:::

. . .

::: {.callout-warning icon=false}
## Fact
Independence is *not* the same as disjointness: two events of positive probability can never be both independent and disjoint.
:::

# The Monty Hall problem

Three doors, one hides a car. You pick a door; the host (who knows where the car is) opens a *different* door revealing a goat. Should you switch?

```python
def monty_hall_trial(switch):
    car_door = rng.integers(0, 3)
    choice = rng.integers(0, 3)
    remaining = [d for d in range(3) if d != choice and d != car_door]
    host_opens = rng.choice(remaining)
    if switch:
        choice = [d for d in range(3) if d != choice and d != host_opens][0]
    return choice == car_door

n_trials = 20000
stay = np.mean([monty_hall_trial(False) for _ in range(n_trials)])
switch = np.mean([monty_hall_trial(True) for _ in range(n_trials)])
print(f"P(win | stay)   ~ {stay:.3f}   (theory: 1/3)")
print(f"P(win | switch) ~ {switch:.3f}   (theory: 2/3)")
```

::: {.callout-caution icon=false}
## Exercise
Use the law of total probability to compute $P(\text{win} \mid \text{switch})$ by conditioning on which door hides the car.
:::

<details><summary>Answer:</summary>
Conditioning on the car's door (each with probability 1/3): if you initially picked the car (prob 1/3), switching loses. If you didn't (prob 2/3), the host's forced reveal means switching always wins. So $P(\text{win}\mid\text{switch}) = \tfrac13\cdot 0 + \tfrac23\cdot 1 = \tfrac23$.
</details>
