# Joint distributions

::: {.callout-note icon=false}
## Definition: Joint and marginal PMF
For discrete $X,Y$: joint PMF $p_{X,Y}(x,y) = P(X=x,Y=y)$; marginals recovered by summing out, $p_X(x) = \sum_y p_{X,Y}(x,y)$.
:::

. . .

::: {.callout-note icon=false}
## Definition: Conditional PMF / independence
$p_{X\mid Y}(x\mid y) = \dfrac{p_{X,Y}(x,y)}{p_Y(y)}$. $X,Y$ are **independent** if $p_{X,Y}(x,y) = p_X(x)p_Y(y)$ for *every* pair $(x,y)$.
:::

---

## Building a joint PMF from simulation

Roll two dice; $X$=sum, $Y$=max. Are $X,Y$ independent?

```python
import pandas as pd
d1 = rng.integers(1, 7, size=500000)
d2 = rng.integers(1, 7, size=500000)
X, Y = d1 + d2, np.maximum(d1, d2)

joint = pd.crosstab(X, Y, normalize=True)
marg_X, marg_Y = joint.sum(axis=1), joint.sum(axis=0)

print("p_{X,Y}(2,1)  =", round(joint.loc[2, 1], 4))
print("p_X(2)*p_Y(1) =", round(marg_X[2] * marg_Y[1], 4))
print("Not equal -> X, Y are NOT independent (the max constrains the sum).")
```

## Exercise
Are the events $\{X = 12\}$ and $\{Y = 6\}$ independent?
:::

<details><summary>Answer:</summary>
$P(X=12)=1/36$, $P(Y=6)=11/36$, and $P(X=12,Y=6)=1/36$ (only $(6,6)$ works). Since $1/36 \neq (1/36)(11/36)$, they are not independent -- but note $P(X=12\mid Y=6) = 1/11 \neq P(X=12)$, consistent.
