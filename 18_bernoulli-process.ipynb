# Stochastic Processes

Stochastic processes are mathematical models of probabilistic experiments that evolve in time and generate a sequence of numerical values.

::: {.callout-note}
## Definition (informal)
A **stochastic process** is a sequence of random variables $X_1, X_2, \dots$ coming from a common experiment (sample space).
:::

**Examples:**

- daily prices of stocks
- scores in a football game
- failure times of a machine
- houly traffic loads at a node in a communication network
- radar measurements of the position of an airplane


## Focus of Stochastic Process Models

- Dependencies among the sequence of values generated by the process
    - e.g., How do future stock prices depend on past values?

. . .

- Long-term averages involving the entire sequence of generated values
    - e.g., What fraction of time is a machine idle?

. . .

- Likelihood/frequency of boundary events
    - e.g., What is the probability that within a given hour all circuits in a telephone system become busy?

## Major Categories of Stochastic Processes

We'll focus on two main types of stochastic processes.

**Arrival-type processes**

- Occurrences that have the character of an "arrival"
- Examples: message receptions, job completions, customer purchases

**Markov processes**

- Experiments that evolve in time and in which the future evolution exhibits a probabilistic dependence on the past
- Examples: daily stock prices, machine failure/repair/replacement

# Bernoulli Process

::: {.callout-note}
## Definition
A **Bernoulli process** is a sequence $X_1, X_2, \dots$ of independent Bernoulli($p$) trials where 
$$\begin{aligned}
& P(X_i = 1) = P(\text{success at the } i \text{th trial}) = p,\\
& P(X_i = 0) = P(\text{failure at the } i \text{th trial}) = 1 - p,
\end{aligned}$$
for each $i$.
:::

. . .

We'll sometimes refer to "successes" as "arrivals".

**Examples:**

- manufacturing defect inspection
- click-through in email ad campaigns
- basketball free throws
- election polling

## Fresh-Start Property

::: {.callout-tip}
## Fact
Given $X_1, \dots, X_n$, the sequence of future trials $X_{n+1}, X_{n+2}, \dots$ are independent Bernoulli trials, and thus form a Bernoulli process independent of the past.
:::

. . .

Starting from any trial, looking only at future trials is a Bernoulli process!

## Memorylessness Property

Let $T$ = time until first success. Then $T \sim \text{Geometric}(p)$:

$$p_T(t) = (1-p)^{t-1}p$$

. . .

The *memorylessness property* says that starting at any point with no prior success, the number of remaining trials until the first success is the same as for a fresh-starting Bernoulli process:

$$P(T = n + t \mid T > n) = (1-p)^{t-1}p = P(T = t)$$

- priority tasks arrive with probability p in each slot, independently, and require the full slot (*busy*)
- nonpriority tasks are always available and executed in slots without priority tasks (*idle*)

**Solution**

$N$ is determined by slots $1, \dots, N$ only, so slots $N+1, N+2, \dots$ form a fresh Bernoulli process, independent of $N$ and of the busy streak that triggered it.

1. $(1-p)^2$
2. Geometric($1-p$): $p_K(k) = p^{k-1}(1-p)$ for $k = 1, 2, \dots$, with mean $\frac{1}{1-p}$


## Interarrival Times

The time of the $k$th success (or arrival) associated with a Bernoulli process is denoted $Y_k$.

. . .

The $k$th interarrival time is a related random variable, denoted by $T_k$.  It is defined as 

$$T_1 = Y_1, \quad \text{ and } T_k = Y_k - Y_{k-1} \quad \text{ for } k = 2, 3, \dots$$

![](figs/interarrival_times.jpg)



## Alternative Description of the Bernoulli Process

1. Start with a sequence of independent geometric random variables $T_1, T_2, \cdots,$ with common parameter $p$, and let these stand for the interarrival times.
2. Record a success (or arrival) at times $T_1, T_1 + T_2, T_1 + T_2 + T_3, \cdots$.

![](figs/interarrival_times.jpg)


## $k$th Arrival Time

The time $Y_k$ of the $k$th success (or arrival) is equal to $$Y_k = T_1 + \cdots + T_k$$ of $k$ independent, identically distributed geometric random variables.

. . .

- Mean and variance:
    - $\mathbb{E}[Y_k] = \mathbb{E}[T_1] + \cdots + \mathbb{E}[T_k] = \frac{k}{p}$
    - $\text{var}(Y_k) = \text{var}(T_1) + \cdots + \text{var}(T_k) = \frac{k(1-p)}{p^2}$
- PMF: $$p_{Y_k}(t) = {t-1 \choose k-1} p^k (1-p)^{t-k}, \quad t = k, k+1, \cdots$$ which is known as the **Pascal PMF of order $k$**.

$$p_{Y_k}(t) = {t-1 \choose k-1} p^k (1-p)^{t-k}, \quad t = k, k+1, \cdots$$

### Example: Basketball Game Fouls

In each minute of a basketball game, Jamie commits a single foul with probability $p$, with fouls in different minutes assumed to be independent.  Jamie will foul out once she commits her fifth foul and otherwise, will play 30 minutes if she does not foul out.  What is the PMF of Jamie's playing time?

. . .

Set it up!  Where is the Bernoulli process with parameter $p$?  What are "successes"?

. . .

Fouls occur as a Bernoulli process with parameter $p$.  Jamie's playing time $Z = \min\{Y_5, 30\}$.  The $5$th arrival time $Y_5$ has PMF $$p_{Y_5}(t) = {t-1 \choose 4} p^5(1-p)^{t-5} \text{ for } t = 5,6,7, \cdots$$

. . .

Then $$p_{Z}(t) = \begin{cases} {z-1 \choose 4} p^5 (1-p)^{z-5} & z = 5,6,7, \cdots, 29 \\ \sum_{t=30}^\infty {t-1 \choose 4}p^5 (1-p)^{t-5} & z = 30 \\ 0 & \text{ otherwise}\end{cases}.$$
