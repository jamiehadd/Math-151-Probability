## Bernoulli Process

::: {.callout-note}
## Definition
A **Bernoulli process** is a sequence $X_1, X_2, \dots$ of independent Bernoulli($p$) trials.
:::

**Examples:**

-

## Fresh-Start Property

::: {.callout-tip}
## Fact
Given $X_1, \dots, X_n$, the sequence of future trials $X_{n+1}, X_{n+2}, \dots$ are independent Bernoulli trials, and thus form a Bernoulli process independent of the past.
:::

---

## Memorylessness Property

Let $T$ = time until first success. Then $T \sim \text{Geometric}(p)$:

$$p_T(t) = (1-p)^{t-1}p$$

. . .

$$P(T = n + t \mid T > n) = (1-p)^{t-1}p = P(T = t)$$
